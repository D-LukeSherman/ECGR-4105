```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Use the URL for the raw CSV data
url = 'https://raw.githubusercontent.com/HamedTabkhi/Intro-to-ML/refs/heads/main/Dataset/D3.csv'

df = pd.read_csv(url)
```

```python
from IPython.display import display
display(df)
```

```python
# Separate features and labels
X1 = df.values[:, 0]  # get input values from first column -- X is a list here
X2 = df.values[:, 1]
X3 = df.values[:, 2]
y = df.values[:, 3]  # get output values from second column -- Y is the list here
m = len(y)  # Number of training examples
n = len(X1)  # Number of training examples

print('m = ', m)
print('n = ', n)
```

```python
X_0 = np.ones((m, 1))
X_1 = X1.reshape(m, 1)
X_2 = X2.reshape(m, 1)
X_3 = X3.reshape(m, 1)
```

```python
# Lets use hstack() function from numpy to stack X_0 and X_1 horizontally (i.e. column
# This will be our final X matrix (feature matrix)
training_1 = np.hstack((X_0, X_1))
training_1[:5]
```

```python
training_2 = np.hstack((X_0, X_2))
training_2[:5]
```

```python
training_3 = np.hstack((X_0, X_3))
training_3[:5]
```

```python
theta = np.zeros(2)
theta
```

```python
def compute_cost(X, y, theta):
    predictions = X.dot(theta)
    errors = np.subtract(predictions, y)
    sqrErrors = np.square(errors)
    J = 1 / (2 * m) * np.sum(sqrErrors)
    return J
```

```python
# Lets compute the cost for theta values
cost_1 = compute_cost(training_1, y, theta)
print('The cost for initial theta values is =', cost_1)
```

```python
def gradient_descent(X, y, theta, alpha, iterations):
    m = len(y)  # Number of training examples
    cost_history = np.zeros(iterations)

    for i in range(iterations):
        predictions = X.dot(theta)
        errors = np.subtract(predictions, y)
        sum_delta = (alpha / m) * X.transpose().dot(errors)
        theta -= sum_delta
        cost_history[i] = compute_cost(X, y, theta)

    return theta, cost_history
```

```python
iterations = 1000
alpha = 0.1
```

```python
theta, cost_history = gradient_descent(training_1, y, theta, alpha, iterations)
print('Final value of theta_0 and theta_1 =', theta)
```

```python
# Assuming that X, y, and theta are already defined
# Also assuming that X has two columns: a feature column and a column of ones

# Scatter plot for the training data
plt.scatter(training_1[:, 1], y, color='red', marker='+', label='Training Data')

# Line plot for the linear regression model
plt.plot(training_1[:, 1], training_1.dot(theta), color='green', label='Linear Regression')

# Plot customizations
plt.rcParams["figure.figsize"] = (10, 6)
plt.grid(True)
plt.xlabel('X_1')
plt.ylabel('Y')
plt.title('Linear Regression Fit')
plt.legend()

# Show the plot
plt.show()
```

```python
plt.plot(range(1, iterations + 1), cost_history, color='blue')
plt.rcParams["figure.figsize"] = (10, 6)
plt.grid(True)

plt.xlabel('Number of iterations')
plt.ylabel('Cost (J)')
plt.title('Convergence of gradient descent')

# Show the plot
plt.show()
```

```python
theta, cost_history = gradient_descent(training_2, y, theta, alpha, iterations)
print('Final value of theta_0 and theta_2 =', theta)
```
